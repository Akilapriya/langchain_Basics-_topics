```python
import os
from dotenv import load_dotenv
load_dotenv()

```

```python
os.environ["OPENAI_API_KEY"] =os.getenv("OPENAI_API_KEY")
os.environ["LANGCHAIN_API_KEY"]=os.getenv('LANGCHAIN_API_KEY')
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.envirn["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI
llm =ChatOpenAI(api_key="OPENAI_API_KEY",model="gpt-4o")
print(llm)
```

```python
result =llm.invoke("what is gen AI?")
result
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt =ChatPromptTemplate.from_messages([
    ("system", "Yow are AI expert. Provide answers based on the questions"),
    ("user", "{input}")
])
prompt
```

```python
chain =prompt|llm
response =chain.invoke({"input": "can you tell me about langchain?"})
response
```

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain= prompt|llm|output_parser
response =chain.invoke({"input": "can you tell me about langchain?"})
response
```
