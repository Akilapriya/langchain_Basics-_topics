```python
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings_1024= OpenAIEmbeddings(model= "text-embeddings-3-large",dimensions =1024)
```

```python
speech=""
with open('speech.txt', 'r') as f:
    speech =f.read()
    
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_docs = text_splitter.create_documents([speech])
```

```python
final_docs
```

```python
pip install chromadb
```

```python
from langchain_community.vectorstores import Chroma
```

```python
db = Chroma.from_documents(final_docs,embeddings_1024)
db
```

```python
query = "it will be easier for us to conduct as belligerent"
retrieved_result =db.similarity_search(query)
print(retrieved_result)
```
